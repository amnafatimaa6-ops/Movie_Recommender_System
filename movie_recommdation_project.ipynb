```python
df = df.drop_duplicates().reset_index(drop = True)
```

```python
#selects only the relevant columns from the original DataFrame.
#it trims the dataset down to key movie features (metadata + popularity metrics)to make analysis cleaner, faster, and more focused.
df = df[['title','overview','genres','tagline','vote_average','popularity']]
df.head()
```

```python
#filling nan in relevant cols
df = df.dropna(subset=['title'])
df['overview'] = df['overview'].fillna('')
df['tagline'] = df['tagline'].fillna('')
df.isnull().sum()
```

```python
#convert 'genres' from stringified list of dicts to a single string of genre names
import ast
df['genres'] = df['genres'].apply(lambda x :''.join([i['name']for i in ast.literal_eval(x)]))
```

```python
df.head()
```

```python
#create a 'tags' column by combining overview, genres, and tagline into a single text string
df['tags']=df['overview']+''+df['genres']+''+df['tagline']
df.head()
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
#download necessary NLTK resources
nltk.download('stopwords') #common words like 'the', 'is', 'and' that usually get removed
nltk.download('wordnet')   #dictionary for lemmatization
```

```python
#prepare tools for text preprocessing
stop_words = set(stopwords.words('english')) #set of words to ignore
lemmatizer = WordNetLemmatizer()             #tool to reduce words to their base form
```

```python
def process_text(text):
    text = str(text).lower()  #converts to lowercase
    text = re.sub(r'[^a-zA-Z\s]', '', text)  #remove punctuation
    words = text.split()
    words = [lemmatizer.lemmatize(word) for word in words if word not in stop_words]  # remove stopwords & lemmatize
    return " ".join(words)  #combine cleaned words back into a single string
```

```python
import ast

def parse_genres(x):
    try:
        #if x is already a list/dict, just use it
        if isinstance(x, list):
            return ' '.join([i['name'] for i in x])
        #if x is a string representing list/dict
        elif isinstance(x, str):
            data = ast.literal_eval(x)
            return ' '.join([i['name'] for i in data])
        else:
            return ''  #empty string for malformed or missing data
    except:
        return ''  #fallback for any error

df['genres'] = df['genres'].apply(parse_genres)

df['tags'] = df['overview'].fillna('') + ' ' + df['genres'].fillna('') + ' ' + df['tagline'].fillna('')
df['tags'] = df['tags'].apply(process_text)
```

```python
#reset index & create indices map ---
df = df.reset_index(drop=True)
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
```

```python
#import the TfidfVectorizer class from scikit-learn
from sklearn.feature_extraction.text import TfidfVectorizer
#keeps only the 5000 most significant words/bigrams to avoid a massive, sparse matrix
#considers both single words and pairs of consecutive words to capture context better
#moves common filler words so only meaningful words affect the model
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1,2), stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['tags']) #for machine learning models or similarity computations
```

```python
from sklearn.metrics.pairwise import cosine_similarity

#recommender function ---
def recommend(title, n=10):
#check if the movie exists in your dataset
    if title not in indices:
        return ['Movie not found']
    #get the index of the movie in your DataFrame
    idx = indices[title]
    #compute cosine similarity between this movie and all movies
    sim_scores = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
    #sort similarity scores in descending order, skip the first one (itself)
    similar_idx = sim_scores.argsort()[::-1][1:n+1]
    #return the titles of the top-n similar movies
    return df['title'].iloc[similar_idx].values
```

```python
#Example usage ---
print(recommend('Toy Story'))
```

```python
print(recommend('The Dark Knight Rises'))
```

```python
!pip install -q sentence-transformers
from sentence_transformers import SentenceTransformer
import numpy as np

#load a pre-trained model (small & fast for now)
model = SentenceTransformer('all-MiniLM-L6-v2')

#encode all movie tags into embeddings
embeddings = model.encode(df['tags'].tolist(), show_progress_bar=True)
```

```python
def semantic_recommend(title, n=10):
    if title not in indices:
        return ['Movie not found']

    idx = indices[title]
    movie_emb = embeddings[idx]

    #compute cosine similarity with all movies
    sim_scores = np.dot(embeddings, movie_emb) / (np.linalg.norm(embeddings, axis=1) * np.linalg.norm(movie_emb))

    #get top n similar movies (excluding itself)
    similar_idx = sim_scores.argsort()[::-1][1:n+1]
    return df['title'].iloc[similar_idx].values
```

```python
print(semantic_recommend('Toy Story'))
print(semantic_recommend('The Dark Knight Rises'))
```

```python
def recommend_by_genre_from_tags(user_genre, n=10):
    user_genre = user_genre.lower().strip()

    #filter movies whose 'tags' contain the genre keyword
    genre_filtered_df = df[df['tags'].str.contains(user_genre, case=False, na=False)]
# - case=False → ignore uppercase/lowercase
    # - na=False → skip rows where 'tags' is missing
    if genre_filtered_df.empty:
        return ['Genre not found']


    return genre_filtered_df['title'].head(n).values

```

```python
print(recommend_by_genre_from_tags('Action'))

```

```python
print(recommend_by_genre_from_tags('Animation'))
```
